```python
def iqr_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    return series[(series < q1 - 1.5 * iqr) | (series > q3 + 1.5 * iqr)]

# Apply IQR outlier detection
gdp_outliers_iqr = iqr_outliers(gdp)
cpi_outliers_iqr = iqr_outliers(cpi)
stock_outliers_iqr = iqr_outliers(stock_data['Close'])
dividends_outliers_iqr = iqr_outliers(dividends)
usoil_outliers_iqr = iqr_outliers(usoil)
ovx_outliers_iqr=iqr_outliers(oilvx_outliers)
wti_outliers_iqr = iqr_outliers(wti)
```

```python
def remove_iqr_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    return series[(series >= Q1 - 1.5 * IQR) & (series <= Q3 + 1.5 * IQR)]
```

```python
gdp_clean = gdp.copy()

# Remove outliers
gdp_clean = remove_iqr_outliers(gdp_clean)

# Remove bad data (e.g., negative or zero)
gdp_clean = gdp_clean[gdp_clean > 0]

# Fill missing values
gdp_clean = gdp_clean.interpolate(method='linear')

gdp_clean = pd.DataFrame(gdp_clean).rename(columns={gdp_clean.name: 'GDP'})


cpi_clean = cpi.copy()

# Remove outliers
cpi_clean = remove_iqr_outliers(cpi_clean)

# Remove bad data (e.g., <= 0)
cpi_clean = cpi_clean[cpi_clean > 0]

# Fill missing values
cpi_clean = cpi_clean.fillna(method='ffill')

cpi_clean = pd.DataFrame(cpi_clean).rename(columns={cpi_clean.name: 'CPI'})

stock_close = stock_data['Close'].copy()

# Remove outliers
stock_close = remove_iqr_outliers(stock_close)

# Remove bad data (e.g., <= 0 or duplicated index)
stock_close = stock_close[stock_close > 0]
stock_close = stock_close[~stock_close.index.duplicated()]

# Fill missing values using rolling mean
stock_close = stock_close.fillna(stock_close.rolling(3, min_periods=1).mean())

stock_clean = pd.DataFrame(stock_close).rename(columns={'Close': 'Stock_Close'})

dividends_clean = dividends.copy()

# Remove outliers
dividends_clean = remove_iqr_outliers(dividends_clean)

# Remove bad data (negative dividends)
dividends_clean = dividends_clean[dividends_clean >= 0]

# Fill missing via backward fill
dividends_clean = dividends_clean.fillna(method='bfill')

dividends_clean = pd.DataFrame(dividends_clean).rename(columns={dividends_clean.name: 'Dividends'})


```

```python
# GDP
if isinstance(gdp_clean, pd.Series):
    gdp_clean = gdp_clean.to_frame(name='GDP')
else:
    gdp_clean.columns = ['GDP']

# CPI
if isinstance(cpi_clean, pd.Series):
    cpi_clean = cpi_clean.to_frame(name='CPI')
else:
    cpi_clean.columns = ['CPI']

# Stock
if 'Close_filled' in stock_data.columns:
    stock_clean = stock_data[['Close_filled']].copy()
    stock_clean.columns = ['CVX_Stock_Close']
else:
    stock_clean.columns = ['CVX_Stock_Close']

# Dividends
if isinstance(dividends_clean, pd.Series):
    dividends_clean = dividends_clean.to_frame(name='Dividends')
else:
    dividends_clean.columns = ['Dividends']

# USOil (you already fixed this to 'USOil_Close')
# usoil_clean already has correct column: 'USOil_Close'
```

```python

# Prepare columns
cols = df2.columns
n = len(cols)
plots_per_row = 3
n_rows = math.ceil(n / plots_per_row)

# Create subplot grid
fig, axes = plt.subplots(n_rows, plots_per_row, figsize=(plots_per_row * 5, n_rows * 4))
axes = axes.flatten()

# Plot KDEs
for i, col in enumerate(cols):
    sns.kdeplot(df2[col].dropna(), fill=True, ax=axes[i])
    axes[i].set_title(f"KDE of {col}")
    axes[i].set_xlabel(col)
    axes[i].grid(True)

# Hide unused axes
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
